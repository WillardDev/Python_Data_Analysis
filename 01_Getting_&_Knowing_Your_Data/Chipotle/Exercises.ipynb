### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 

### Step 3. Assign it to a variable called chipo.

```python
chipo = pd.read_csv('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv', sep='\t')
```

### Step 4. See the first 10 entries

```python
print(chipo.head(10))
```

### Step 5. What is the number of observations in the dataset?

```python
# Solution 1
print(len(chipo))
```

### Step 6. What is the number of columns in the dataset?

```python
print(len(chipo.columns))
```

### Step 7. Print the name of all the columns.

```python
print(chipo.columns.tolist())
```

### Step 8. How is the dataset indexed?

```python
print(chipo.index)
```

### Step 9. Which was the most-ordered item? 

```python
print(chipo.groupby('item_name')['quantity'].sum().idxmax())
```

### Step 10. For the most-ordered item, how many items were ordered?

```python
print(chipo.groupby('item_name')['quantity'].sum().max())
```

### Step 11. What was the most ordered item in the choice_description column?

```python
print(chipo['choice_description'].value_counts().idxmax())
```

### Step 12. How many items were orderd in total?

```python
print(chipo['quantity'].sum())
```

### Step 13. Turn the item price into a float

#### Step 13.a. Check the item price type

```python
print(chipo['item_price'].dtype)
```

#### Step 13.b. Create a lambda function and change the type of item price

```python
chipo['item_price'] = chipo['item_price'].apply(lambda x: float(x[1:]))
chipo['item_price']
```

#### Step 13.c. Check the item price type

```python
print(chipo['item_price'].dtype)
```

### Step 14. How much was the revenue for the period in the dataset?

```python
chipo['revenue'] = chipo['quantity'] * chipo['item_price']
print(chipo['revenue'].sum())
```

### Step 15. How many orders were made in the period?

```python
print(chipo['order_id'].nunique())
```

### Step 16. What is the average revenue amount per order?

```python
# Solution 1
print(chipo.groupby('order_id')['revenue'].sum().mean())
```

### Step 17. How many different items are sold?

```python
print(chipo['item_name'].nunique())
```
