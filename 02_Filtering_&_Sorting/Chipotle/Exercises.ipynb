### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 

### Step 3. Assign it to a variable called chipo.

```python
chipo = pd.read_csv('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv', sep='\t')
chipo.head()
```

### Step 4. How many products cost more than $10.00?

```python
chipo['item_price'] = chipo['item_price'].str.replace('$', '').astype(float)

products_over_10 = len(chipo[chipo['item_price'] > 10])
products_over_10
```

### Step 5. What is the price of each item? 
###### print a data frame with only two columns item_name and item_price

```python
price_df = chipo[['item_name', 'item_price']]
price_df_unique = price_df.drop_duplicates()
price_df_unique
```

### Step 6. Sort by the name of the item

```python
sorted_items = price_df_unique.sort_values('item_name')
sorted_items.head(15)
```

### Step 7. What was the quantity of the most expensive item ordered?

```python
most_expensive_idx = chipo['item_price'].idxmax()
most_expensive_row = chipo.loc[most_expensive_idx]
quantity = most_expensive_row['quantity']
quantity
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
veggie_count = chipo[chipo['item_name'] == 'Veggie Salad Bowl']['quantity'].sum()
veggie_count
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
canned_soda_multi = chipo[(chipo['item_name'] == 'Canned Soda') & (chipo['quantity'] > 1)]
count = len(canned_soda_multi)
count
```
