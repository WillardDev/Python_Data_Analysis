### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user). 

### Step 3. Assign it to a variable called users.

```python
users = pd.read_csv(
    'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user',
    sep='|',
    names=['user_id', 'age', 'gender', 'occupation', 'zip_code']
)

users.head()
```

### Step 4. Discover what is the mean age per occupation

```python
users['age'] = pd.to_numeric(users['age'], errors='coerce')

mean_age_by_occupation = users.groupby('occupation')['age'].mean()

print(mean_age_by_occupation)
print(f"\nType: {type(mean_age_by_occupation)}")
print("\nSorted from oldest to youngest:")
print(mean_age_by_occupation.sort_values(ascending=False))
```

### Step 5. Discover the Male ratio per occupation and sort it from the most to the least

```python
male_count = users[users['gender'] == 'M'].groupby('occupation').size()
total_count = users.groupby('occupation').size()
male_ratio = male_count / total_count

male_ratio_sorted = male_ratio.sort_values(ascending=False)
print(male_ratio_sorted)
print(f"\nType: {type(male_ratio_sorted)}")
```

### Step 6. For each occupation, calculate the minimum and maximum ages

```python
age_range = users.groupby('occupation')['age'].agg(['min', 'max'])
print(age_range)
print(f"\nType: {type(age_range)}")
```

### Step 7. For each combination of occupation and gender, calculate the mean age

```python
mean_age_by_occ_gender = users.groupby(['occupation', 'gender'])['age'].mean()
print(mean_age_by_occ_gender)
print(f"\nType: {type(mean_age_by_occ_gender)}")

print("\n" + "="*60)
print("As a DataFrame:")
print("="*60)
mean_age_by_occ_gender_df = mean_age_by_occ_gender.unstack()
print(mean_age_by_occ_gender_df)
```

### Step 8.  For each occupation present the percentage of women and men

```python
total_by_occ = users.groupby('occupation').size()
male_by_occ = users[users['gender'] == 'M'].groupby('occupation').size()
female_by_occ = users[users['gender'] == 'F'].groupby('occupation').size()

# Create a DataFrame with the results
gender_percentage = pd.DataFrame({
    'male_pct': (male_by_occ / total_by_occ * 100).round(2),
    'female_pct': (female_by_occ / total_by_occ * 100).round(2)
})

print(gender_percentage)
print(f"\nType: {type(gender_percentage)}")
```
