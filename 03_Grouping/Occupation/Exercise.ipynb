### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user). 

### Step 3. Assign it to a variable called users.

```python
users = pd.read_csv(
    'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user',
    sep='|',
    names=['user_id', 'age', 'gender', 'occupation', 'zip_code']
)

users.head()
```

### Step 4. Discover what is the mean age per occupation

```python
# Step 1: Calculate mean age for each occupation
mean_age_by_occupation = users.groupby('occupation')['age'].mean()

# Step 2: See the data and type
print(mean_age_by_occupation)
print(f"\nType: {type(mean_age_by_occupation)}")

# Step 3: Sort from oldest to youngest professions
print("\nSorted from oldest to youngest:")
print(mean_age_by_occupation.sort_values(ascending=False))
```

### Step 5. Discover the Male ratio per occupation and sort it from the most to the least

```python
male_count = users[users['gender'] == 'M'].groupby('occupation').size()
total_count = users.groupby('occupation').size()
male_ratio = male_count / total_count

male_ratio_sorted = male_ratio.sort_values(ascending=False)
print(male_ratio_sorted)
print(f"\nType: {type(male_ratio_sorted)}")
```
