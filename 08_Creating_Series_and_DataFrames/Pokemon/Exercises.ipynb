### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
print(pd.__version__)
print(np.__version__)
```

### Step 2. Create a data dictionary that looks like the DataFrame below

```python
# create a data dictionary that looks like the DataFrame below
data = {
        'evolution': ['Ivysaur', 'Charmeleon', 'Wartortle', 'Metapod'],
        'hp': [45, 39, 44, 45],
        'name': ['Bulbasaur', 'Charmander', 'Squirtle', 'Caterpie'],
        'pokedex': ['yes', 'no', 'yes', 'no'],
        'type': ['grass', 'fire', 'water', 'bug'],
}
```

### Step 3. Assign it to a variable called pokemon

```python
### Step 3. Create a DataFrame called pokemon
pokemon = pd.DataFrame(data)
pokemon
```

### Step 4. Ops...it seems the DataFrame columns are in alphabetical order. Place  the order of the columns as name, type, hp, evolution, pokedex

```python
# Reorder columns
pokemon = pokemon[['name', 'type', 'hp', 'evolution', 'pokedex']]
pokemon
```

### Step 5. Add another column called place, and insert what you have in mind.

```python
# Add a new column 'place' with Pokemon habitat locations
pokemon['place'] = ['Forest', 'Mountain', 'Ocean', 'Forest']
pokemon
```

### Step 6. Present the type of each column

```python
# Display the data type of each column
print(pokemon.dtypes)
```

### BONUS: Question - Which Pokemon are in the Pokedex and have the highest HP?

```python
answer = pokemon[pokemon['pokedex'] == 'yes'].sort_values('hp', ascending=False)
print("Pokemon in Pokedex sorted by HP (highest first):")
print(answer[['name', 'type', 'hp']])
```
