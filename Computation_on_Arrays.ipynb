## Computation on NumPy Arrays: Universal Functions

```python
import numpy as np
```

```python
x = np.array([2, 3, 4])
y = np.array([7, 8, 9])
# x + y
```

```python
xy = np.add(x, y)
xy
```

### Slow: Using Python Loops instead of vectorization

```python
x = [2, 3, 4]
y = [7, 8, 9]
x + y
```

```python
z = []
for i in range(len(x)):
    z.append(x[i]+y[i])
z
```

```python
# define two sample arrays
a = np.array([10, 20, 30])
b = np.array([3, 4, 5])

```

```python
# addition of two arrays
c = np.add(a, b)
c
# alternative
# c = a + b
```

```python
# subtraction of two arrays
d = np.subtract(a, b)
d
# alternative
# d = a - b
```

```python
# multiplication of two arrays
e = np.multiply(a, b)
e
# alternative
# e = a * b
```

```python
# division of two arrays
f = np.divide(a, b)
f
# alternative
# f = a / b
```

```python
# floor division of two arrays
g = np.floor_divide(a, b)
g
# alternative
# g = a // b
```

```python
# unary operations on arrays
h = np.negative(a)
h
```
