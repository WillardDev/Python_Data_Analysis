```python
# define two sample arrays
a = np.array([10, 20, 30])
b = np.array([3, 4, 5])

```

```python
# addition of two arrays
c = np.add(a, b)
c
# alternative
# c = a + b
```

```python
# subtraction of two arrays
d = np.subtract(a, b)
d
# alternative
# d = a - b
```

```python
# multiplication of two arrays
e = np.multiply(a, b)
e
# alternative
# e = a * b
```

```python
# division of two arrays
f = np.divide(a, b)
f
# alternative
# f = a / b
```

```python
# floor division of two arrays
g = np.floor_divide(a, b)
g
# alternative
# g = a // b
```

```python
# unary operations on arrays
h = np.negative(a)
h
# alternative
# h = -a
```

```python
# power of two arrays
i = np.power(a, b)
i
# alternative
# i = a ** b
```

```python
# modulus of two arrays
j = np.mod(a, b)
j
# alternative
# j = a % b
```

## Broadcasting in NumPy

```python
# adding a scalar to an array
a = np.array([1, 2, 3])
b = 5
c = a + b
c
```

```python
# add 1D to a 2D array
a = np.array([[1, 2, 3], [4, 5, 6]])
print('a:', a)
print('a shape:', a.shape)
b = np.array([10, 20, 30])
print('b:', b)
print('b shape:', b.shape)
c = a + b
print('c:', c)
print('c shape:', c.shape)
```

```python
a = np.arange(3).reshape(3, 1)
b = np.arange(3)
print('Shape A:', a.shape)
print('Shape B:', b.shape)
c = a + b
print('Shape C:', c.shape)
```

```python
print(c)
```

```python
# arrays that are not compatible for broadcasting
M = np.array((3, 2), dtype=int)
a = np.arange(3)
# a + M
```

```python
# more examples of arrays that are not compatible for broadcasting
a = np.arange(3).reshape(3, 1)
print('Shape A:', a.shape)
b= np.zeros((2, 2))
print('Shape B:', b.shape)
# a + b
```

## Boolean Masking

```python
arr = np.array([10, 20, 30, 40, 50])

# create the mask
mask = arr > 25

# using the mask
result = arr[mask]
print('Array:', arr)
print('Mask:', mask)
print('Result:', result)
```

```python
arr = np.arange(1, 11)
mask = arr > 5
result = arr[mask]
print('Array:', arr)
print('Mask:', mask)
print('Result:', result)
```

### Boolean masking with 2D arrays

```python
arr2D = np.arange(1, 10).reshape(3, 3)
print('2D Array:\n', arr2D)
```

```python
# get all elements greater than 5
mask2D = arr2D > 5
print('Mask 2D:\n', mask2D)
```

```python
result = arr2D[mask2D]
print('Result:\n', result)
```

```python
arr2D[arr2D > 5] = 0
print('2D Array:\n', arr2D)
```

## Counting Entries

```python
# check how many values are gonna be less than 6
x = np.random.randint(1, 13, size=12).reshape(3, 4)
print('Array:', x)
mask = x < 6
print('Mask:', mask)
nz = np.count_nonzero(x < 6)
print('Number of values less than 6:', nz)
```

### Boolean Operators

```python
y = np.array([10, 15, 20, 25, 30])
print('Array:', y)
```

```python
z = y < 15
print('Mask:', z)
```

```python
w = y > 25
print('Mask:', w)
```

```python
mask = z | w 
print('Mask:', mask)
```

```python
result = y[mask]
print('Result:', result)
```

```python
v = ~(y > 20)
print("V:", v)
```

```python
result_not = y[v]
print('Result Not:', result_not)
```
