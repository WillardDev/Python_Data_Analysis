## Computation on NumPy Arrays: Universal Functions

```python
import numpy as np
```

```python
x = np.array([2, 3, 4])
y = np.array([7, 8, 9])
# x + y
```

```python
xy = np.add(x, y)
xy
```

### Slow: Using Python Loops instead of vectorization

```python
x = [2, 3, 4]
y = [7, 8, 9]
x + y
```

```python
z = []
for i in range(len(x)):
    z.append(x[i]+y[i])
z
```

```python
# define two sample arrays
a = np.array([10, 20, 30])
b = np.array([3, 4, 5])

```

```python
# addition of two arrays
c = np.add(a, b)
c
# alternative
# c = a + b
```

```python
# subtraction of two arrays
d = np.subtract(a, b)
d
# alternative
# d = a - b
```

```python
# multiplication of two arrays
e = np.multiply(a, b)
e
# alternative
# e = a * b
```

```python
# division of two arrays
f = np.divide(a, b)
f
# alternative
# f = a / b
```

```python
# floor division of two arrays
g = np.floor_divide(a, b)
g
# alternative
# g = a // b
```

```python
# unary operations on arrays
h = np.negative(a)
h
# alternative
# h = -a
```

```python
# power of two arrays
i = np.power(a, b)
i
# alternative
# i = a ** b
```

```python
# modulus of two arrays
j = np.mod(a, b)
j
# alternative
# j = a % b
```

## Aggregation in Numpy

### Summing the values in an array

```python
R = np.arange(1, 11)
R
```

```python
sum_R = sum(R)
sum_R
# alternative
# sum_R = np.sum(R)
```

```python
R2 = np.random.random(100)
R2
```

```python
# using python sum function
sum_R2 = sum(R2)
sum_R2
```

```python
# using numpy to sum the values in an array
sum_R2 = np.sum(R2)
sum_R2
```

```python
%timeit sum(R2)
```

```python
%timeit np.sum(R2)
```

```python
R3 = np.random.random(1000000)
R3
```

```python
%timeit np.sum(R3)
```

```python
%timeit sum(R3)
```

```python
%timeit np.max(R3)
```

```python
%timeit max(R3)
```

```python
M.sum()
```

## Broadcasting in NumPy

```python
# adding a scalar to an array
a = np.array([1, 2, 3])
b = 5
c = a + b
c
```

```python
# add 1D to a 2D array
a = np.array([[1, 2, 3], [4, 5, 6]])
print('a:', a)
print('a shape:', a.shape)
b = np.array([10, 20, 30])
print('b:', b)
print('b shape:', b.shape)
c = a + b
print('c:', c)
print('c shape:', c.shape)
```
