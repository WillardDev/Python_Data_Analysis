## Series as specialized Dictionaries

```python
student_dict = {
    'name': 'Alice',
    'age': 21,
    'city': 'New York'
}

student_series = pd.Series(student_dict)
student_series
```

```python
population_dict = {
    'California': 38332521,
    'Texas': 26448193,
    'New York': 19651127,
    'Florida': 19552860,
    'Illinois': 12882135
}
population_series = pd.Series(population_dict)
population_series
```

```python
# explicit index
population_series['Texas':'Florida']
```

```python
# implicit index
population_series[1:4]
```
