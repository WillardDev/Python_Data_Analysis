## Series as specialized Dictionaries

```python
student_dict = {
    'name': 'Alice',
    'age': 21,
    'city': 'New York'
}

student_series = pd.Series(student_dict)
student_series
```

```python
population_dict = {
    'California': 38332521,
    'Texas': 26448193,
    'New York': 19651127,
    'Florida': 19552860,
    'Illinois': 12882135
}
population_series = pd.Series(population_dict)
population_series
```

```python
# explicit index
population_series['Texas':'Florida']
```

```python
# implicit index
population_series[1:4]
```

## The Pandas DataFrame Object

### DataFrame as a generalized NumPy array

```python
area_dict = {
    'California': 423967, 
    'Texas': 695662, 
    'New York': 141297,
    'Florida': 170312, 
    'Illinois': 149995
}

area_series = pd.Series(area_dict)
area_series
```

```python
states = pd.DataFrame({
    'population': population_series,
    'area': area_series
})
states
```

```python
states.population['California']
```

```python
print('New York Population:',states.population['New York'])
```

```python
print('Illinois Area:',states.area['Illinois'])
```

```python
# slicing
# states.population[0:3]
states.population['California':'New York']
```

## DataFrame as a specialized dictionary

### From a single Series Object

```python
pd.DataFrame(population_series, columns=['population'])
```

### From a list of dicts

```python
# create a DataFrame from a list of dicts
states = pd.DataFrame(data)
states
```
