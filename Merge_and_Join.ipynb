# Merge and Join

```python
import pandas as pd
import numpy as np
```

## Categories of Join

### One-to-One Joins

```python
df1 = pd.DataFrame({
    'employee': ['Mercy', 'Mary', 'Mark', 'Maureen'],
    'departments': ['Accounting', 'Engineering', 'Engineering', 'HR'],
})

df2 = pd.DataFrame({
    'employee': ['Mercy', 'Mary', 'Mark', 'Maureen'],
    'hire_date': [2004, 2006, 2008, 2010],
})

display(df1, df2)
```

```python
df3 = pd.merge(df1, df2, on='employee')
df3
```

### Many-to-One Joins

```python
df4 = pd.DataFrame({
    'departments': ['Accounting', 'Engineering', 'HR'],
    'supervisor': ['Carly', 'Guido', 'Steve'],
})

display(df3, df4)
```
