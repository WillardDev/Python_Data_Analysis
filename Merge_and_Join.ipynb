# Merge and Join

```python
import pandas as pd
import numpy as np
```

## Categories of Join

### One-to-One Joins

```python
df1 = pd.DataFrame({
    'employee': ['Mercy', 'Mary', 'Mark', 'Maureen'],
    'departments': ['Accounting', 'Engineering', 'Engineering', 'HR'],
})

df2 = pd.DataFrame({
    'employee': ['Mercy', 'Mary', 'Mark', 'Maureen'],
    'hire_date': [2004, 2006, 2008, 2010],
})

display(df1, df2)
```

```python
df3 = pd.merge(df1, df2, on='employee')
df3
```

### Many-to-One Joins

```python
df4 = pd.DataFrame({
    'departments': ['Accounting', 'Engineering', 'Engineering', 'HR'],
    'supervisor': ['Carly', 'Guido', 'Steve', 'Melody'],
})

display(df3, df4)
```

```python
df5 = pd.merge(df3, df4)
df5
```

### Many-to-Many Joins

```python
df6 = pd.DataFrame({
    'departments': ['Accounting', 'Accounting', 'Engineering', 'Engineering', 'HR', 'HR'],
    'skills': ['Math', 'Software', 'Presentation', 'Math', 'Communication', 'Spreadsheet'],
})

display(df4, df6, pd.merge(df4, df6))
```

## Specification of the Merge Key

```python
display(df1, df2)
```

```python
display(df1, df2, pd.merge(df1, df2, on='employee'))
```

### The left_on and right_on Keywords

```python
df3 = pd.DataFrame({
    'name': ['Bob', 'Mercy', 'Mary', 'Mark'],
    'salary': [70000, 10000, 50000, 30000]
})

display(df1, df3, pd.merge(df1, df3, left_on="employee", right_on="name"))
```

#### Using drop() to drop desired columns

```python
display(df1, df3, pd.merge(df1, df3, left_on="employee", right_on="name").drop('name', axis=1))
```

### The left_index and right_index Keywords

```python
df1a = df1.set_index('employee')
df2a = df2.set_index('employee')

display(df1a, df2a)
```

```python
display(df1a, df2a, pd.merge(df1a, df2a, left_index=True, right_index=True))
```

```python
df1a.join(df2a)
```

```python
display(df1a, df3, pd.merge(df1a, df3, left_index=True, right_on='name'))
```

## Specifying Set Arithmetic for Joins

```python
df6 = pd.DataFrame({
    'names': ['Peter', 'Mary', 'Paul'],
    'food': ['fish', 'beans', 'bread']
})

df7 = pd.DataFrame({
    'names': ['Mary', 'Paul'],
    'drink': ['wine', 'beer']
})
```

```python
display(df6, df7, pd.merge(df6, df7))
```

```python
pd.merge(df6, df7, how='inner')
```

```python
display(df6, df7, pd.merge(df6, df7, how='outer'))
```

```python
display(df6, df7, pd.merge(df6, df7, how='left'))
```

```python
display(df6, df7, pd.merge(df6, df7, how='right'))
```

### Overlapping Column Names

```python
df8 = pd.DataFrame({
    'names': ['Peter', 'Mary', 'Paul'],
    'food': ['fish', 'beans', 'bread']
})

df9 = pd.DataFrame({
    'names': ['Mary', 'Paul'],
    'food': ['fish', 'beans']
})
```

```python
display(df8, df9, pd.merge(df8, df9, on='names'))
```
