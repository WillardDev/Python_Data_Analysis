#### 4. How to find the memory size of any array (★☆☆)

```python
arr = np.array([1, 2, 3, 4, 5])
memory_size = arr.size * arr.itemsize
print(f"Memory size of array: {memory_size} bytes")
```

#### 6. Create a null vector of size 10 but the fifth value which is 1 (★☆☆)

```python
null_vector_fifth = np.zeros(10)
null_vector_fifth[4] = 1  # index 4 is the 5th element
print("Null vector with fifth element = 1:", null_vector_fifth)
```

#### 7. Create a vector with values ranging from 10 to 49 (★☆☆)

```python
vector_range = np.arange(10, 50)
print("Vector from 10 to 49:", vector_range)
```

#### 8. Reverse a vector (first element becomes last) (★☆☆)

```python
vector = np.arange(10, 20)
reversed_vector = vector[::-1]
print("Original vector:", vector)
print("Reversed vector:", reversed_vector)
```

#### 9. Create a 3x3 matrix with values ranging from 0 to 8 (★☆☆)

#### 10. Find indices of non-zero elements from [1,2,0,0,4,0] (★☆☆)

```python
arr = np.array([1, 2, 0, 0, 4, 0])
non_zero_indices = np.nonzero(arr)
print("Array:", arr)
print("Indices of non-zero elements:", non_zero_indices[0])
```

#### 12. Create a 3x3x3 array with random values (★☆☆)

```python
random_3d = np.random.random((3, 3, 3))
print(random_3d)
```

#### 13. Create a 10x10 array with random values and find the minimum and maximum values (★☆☆)

```python
np.random.seed(42) 
random_10x10 = np.random.random((10, 10))
min_value = random_10x10.min()
max_value = random_10x10.max()
print("10x10 random array:")
print(random_10x10)
print(f"Minimum value: {min_value}")
print(f"Maximum value: {max_value}")
```

#### 14. Create a random vector of size 30 and find the mean value (★☆☆)

```python
np.random.seed(42)
random_vector_30 = np.random.random(30)
mean_value = random_vector_30.mean()
print("Random vector of size 30:", random_vector_30)
print(f"Mean value: {mean_value}")
```

#### 15. Create a 2d array with 1 on the border and 0 inside (★☆☆)

```python
border_array = np.ones((5, 5))
border_array[1:-1, 1:-1] = 0
print("2D array with 1 on border and 0 inside:")
print(border_array)
```
