#### 33. How to get the dates of yesterday, today and tomorrow? (★☆☆)

````python
# Using NumPy datetime
today = np.datetime64('today', 'D')
yesterday = today - np.timedelta64(1, 'D')
tomorrow = today + np.timedelta64(1, 'D')

print("NumPy dates:")
print("Yesterday:", yesterday)
print("Today:", today)
print("Tomorrow:", tomorrow)

# Using Python datetime
from datetime import datetime, timedelta
today_py = datetime.now().date()
yesterday_py = today_py - timedelta(days=1)
tomorrow_py = today_py + timedelta(days=1)

print("\nPython datetime:")
print("Yesterday:", yesterday_py)
print("Today:", today_py)
print("Tomorrow:", tomorrow_py)
````

#### 35. How to compute ((A+B)*(-A/2)) in place (without copy)? (★★☆)

````python
A = np.array([1, 2, 3, 4, 5], dtype=float)
B = np.array([5, 4, 3, 2, 1], dtype=float)

print("A =", A)
print("B =", B)

# Compute in place: ((A+B)*(-A/2))
# Using np.add, np.multiply, np.negative
C = A.copy()
np.add(A, B, out=C)      # C = A + B
np.negative(A, out=A)    # A = -A (in place)
np.multiply(A, 0.5, out=A)  # A = A * 0.5 = -A/2
np.multiply(C, A, out=C)  # C = C * A = (A+B)*(-A/2)

print("\nResult:", C)
````
