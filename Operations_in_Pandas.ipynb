## Ufuncs: Index Alignment

### Index Alignment in Series

```python
area = pd.Series({
            'Alaska': 1723337, 
            'Texas': 695662,
            'California': 423967
        }, 
        name='area')

population = pd.Series({
                'California': 39538223, 
                'Texas': 29145505,
                'Florida': 21538187
            }, 
            name='population')
print('Area:', area)
print('----------------')
print('Population:', population)
```

```python
population/area
```

```python
area.index.union(population.index)
```

```python
A = pd.Series([2, 4, 6], index=[0, 1, 2])
B = pd.Series([1, 3, 5], index=[1, 2, 3])
print(A)
print(B)
```

```python
addition = A + B
addition
```

```python
A.add(B, fill_value=0)
```

```python
C = pd.DataFrame(np.random.randint(1, 20, (2, 2)), columns=['A', 'B'])
C
```

```python
D = pd.DataFrame(np.random.randint(0, 10, (3, 3)), columns=['A', 'B', 'C'])
D
```

## Ufuncs: Operations Between DataFrame and Series

```python
df = pd.DataFrame(A, columns=['A', 'B', 'C', 'D'])
df
```
