## Ufuncs: Index Alignment

### Index Alignment in Series

```python
area = pd.Series({
            'Alaska': 1723337, 
            'Texas': 695662,
            'California': 423967
        }, 
        name='area')

population = pd.Series({
                'California': 39538223, 
                'Texas': 29145505,
                'Florida': 21538187
            }, 
            name='population')
print('Area:', area)
print('----------------')
print('Population:', population)
```

```python
population/area
```
