## Creating Arrays from Scratch

```python
# create an arary of zeros
np.zeros(5, dtype='int')
```

```python
# create a 3x5 filled in with ones
np.ones((3,5), dtype='int')
```

```python
# create a 3x5 filled with 3.14
np.full((3,5), 3.14)
```

```python
# create an aray fiiled in with a linear sequence
np.arange(0, 20, 2)
```

```python
# create an array of five values evenly spaced between 0 and 1
np.linspace(0, 1, 5)
```

```python
# create a 3x3 array of uniformly distributed random values between 0 and 1
np.random.random((3,3))
```

```python
# create a 3x3 array of nomarlly distributed random values with mean 0 and standard deviation 1
np.random.normal(0, 1, (3,3))
```

```python
# create a 3x3 array of random integers between 0 and 10
np.random.randint(0, 10, (3,3))
```

```python
# create a 3x3 identity matrix
np.eye(3)
```

```python
# create an uninitialized array of three integers
# the values will be whatever happens to already exist at that memory location
np.empty(3, dtype='int')
```

## Data Manipulation

### NumPy Array Attributes

```python
np.random.seed(42) # set the random seed for reproducibility

# create a one-dimensional array
x1 = np.random.randint(10, size=6)
x1

# create a two-dimensional array
x2 = np.random.randint(10, size=(3,4))
x2

# create a three-dimensional array
x3 = np.random.randint(10, size=(3,4,5))
x3
```

```python
print("x1:", x1)
print("x2: ", x2)
print("x3:", x3)
```

```python
print("x1 shape:", x1.shape)
print("x2 shape:", x2.shape)
print("x3 shape:", x3.shape)
```

```python
print("x1 ndim: ", x1.ndim)
print("x2 ndim: ", x2.ndim)
print("x3 ndim: ", x3.ndim)
```

```python
print('x1 size: ', x1.size, 'bytes')
print('x2 size: ', x2.size, 'bytes')
print('x3 size: ', x3.size, 'bytes')
```

```python
print('x1 dtype: ', x1.dtype)
print('x2 dtype: ', x2.dtype)
print('x3 dtype: ', x3.dtype)
```
